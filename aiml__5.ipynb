```python
import matplotlib.pyplot as plt
```

```python
# Define the missing data for x and y
x = [1, 2, 3, 4, 5]
y = [85, 90, 78, 92, 88]

plt.plot(x, y)
plt.xlabel("Roll no")
plt.ylabel("Marks")
plt.title("Marksheet")
plt.show()
```

```python
#marker and line styles
plt.plot(x,y,marker='o', linestyle='--',color ='r')
plt.show()
```

```python
#multiple lines

x = [1,2,3,4,5]
y1= [10 , 20 , 15 ,25, 30]
y2= [5,15, 20 , 22,35]

plt.plot (x,y1, label ="Dataset 1")
plt.plot (x,y2, label ="Dataset 2")

plt.xlabel("X")
plt.ylabel("Values")
plt.title("comparision of Datasets")
plt.legend()
plt.grid()
plt.show()
```

```python
#Scatter plot

hours = [1,2,3,4,5,6]
marks = [35 , 42 , 50 , 55 , 65 , 72]

plt.scatter(hours , marks)
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.title("Hours vs Marks")
plt.show()
```

```python
# histograms

marks = [35 , 42 , 50 , 55 , 65 , 70 , 90 , 92 , 52]

plt.hist(marks , bins = 5)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()
```

#Bar chart

```python
subjects = ("python", "Pl", "DBMS", "AI")
MARKS = (80 , 78 , 92 , 89)

plt.bar(subjects , MARKS)
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Marks in Subjects")
plt.show()
```

#Sub Plot

```python
plt.subplot(1,2,1)
plt.plot(x,y1)
plt.title("Line Plot")

plt.subplot(1,2,2)
plt.scatter(x,y2)
plt.title("Scatter Plot")

plt.show()
```

# Pie Chart

```python
subjects = ("python", "Pl", "DBMS", "AI")
MARKS = (80 , 78 , 92 , 89)

plt.pie(MARKS , labels = subjects)
plt.show()
```
